```python
import pandas as pd 
import os 

RAW_DATA_DIR = "../data/raw"

tickers = ["AAPL", "MSFT", "GOOGL", "AMZN"]

dfs = []
for ticker in tickers:
    file_path = os.path.join(RAW_DATA_DIR,ticker +"_1d.csv")
    df = pd.read_csv(file_path)
    display(df.head())
    df['Date'] = pd.to_datetime(df['Date'])
    df = df.set_index('Date')
    display(df.head())
    dfs.append(df)
```

```python
from ipywidgets import interact

import matplotlib.pyplot as plt
def show_raw_data(i):
    df = dfs[i]
    ticker = tickers[i]
    plt.figure(figsize=(14, 6))
    plt.plot(df['Close'], label="Close Price", linewidth=1.2)
    plt.title(f"{tickers[i]} Close Price Over Time", fontsize=16)
    plt.xlabel("Date")
    plt.ylabel("Price")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()


interact(show_raw_data, i=(0, len(tickers)-1))
```
